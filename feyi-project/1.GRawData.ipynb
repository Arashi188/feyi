# HOW TO GENERATE RAW DATA

```python
# Step1:create a dictionary

data2 = {
    "Employee_ID":[],
    "Employee_Name":[],
    "Department":[],
    "Region":[],
    "Product":[],
    "Units_Sold":[],
    "units_price":[],
    "Revenue":[],
    "Target":[],
    "Year":[]
}

# Create Lists
department = ["Sales", "finance", "HR", "IT", "Procurement"]
region = ["North", "South", "East", "West"]
product = ["Product A", "Product B", "Product C", "Product D", "Product E"]

# create for loop
for i in range(1,501):
    unit = random.randint(10,500)
    price = random.randint(100,1000)
    revenue = unit * price
    
    data2["Employee_ID"].append(f"EMP- {i}"),
    data2["Employee_Name"].append(f"Employee_{i}"),
    data2["Department"].append(random.choice(department)),
    data2["Region"].append(random.choice(region)),
    data2[ "Product"].append(random.choice(product)),
    data2["Units_Sold"].append(unit),
    data2[ "units_price"].append(price),
    data2[ "Revenue"].append(revenue),
    data2["Target"].append(random.randint(100000,500000)),
    data2["Year"].append(random.choice([2024,2025,2026]))



# Create Dataframe
df = pd.DataFrame(data2)
display(df.head())
display(df.tail())

```

#  Step 2: Save and Read Dataset


```python
# Save your Random Data into a csv file
df.to_csv("company_data2.csv", index=False)
```
